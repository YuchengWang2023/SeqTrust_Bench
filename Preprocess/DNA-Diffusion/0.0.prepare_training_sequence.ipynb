```python
import pandas as pd
```

```python
def filter_specific_label(df, token_cols, threshold=2):

    sub_df = df.copy()

    tokens = (
        sub_df[token_cols]
        .astype(int)
        .values
    )

    max_token = tokens.max(axis=1)

    second_token = (
        tokens.sum(axis=1)
        - tokens.max(axis=1)
        - tokens.min(axis=1)
    )

    sub_df["TAG"] = (
        sub_df[token_cols]
        .astype(int)
        .idxmax(axis=1)
        .str.replace("_token$", "", regex=True)
    )

    sub_df = sub_df[
        max_token - second_token >= threshold
    ]

    return sub_df.reset_index(drop=True)
```

# DNase

```python
DNase_complete = pd.read_csv("/home/wangyucheng/projects/SeqEval_Bench_Data/DNase/CellType/DNase_complete_starting_sequences_regLM.tsv", sep="\t")
DNase_offline = pd.read_csv("/home/wangyucheng/projects/SeqEval_Bench_Data/DNase/CellType/DNase_offline_starting_sequences_regLM.tsv", sep="\t")
```

```python
cell_types = ["GM12878", "HepG2", "K562"]
prediction_cols = [f"functional_score_{x}" for x in cell_types]
token_cols = [f"{x}_token" for x in prediction_cols]

# The DNADiffusion dataloader requires lowercase "sequence" / "chr" columns
DNase_complete = DNase_complete.rename(columns={"Sequence": "sequence", "chrom": "chr"})
DNase_offline = DNase_offline.rename(columns={"Sequence": "sequence", "chrom": "chr"})

DNase_complete_DNADiffusion = filter_specific_label(DNase_complete.copy(), token_cols=token_cols, threshold=2)
DNase_offline_DNADiffusion = filter_specific_label(DNase_offline.copy(), token_cols=token_cols, threshold=2)

print(f"complete: {len(DNase_complete)} -> {len(DNase_complete_DNADiffusion)}")
print(f"offline:  {len(DNase_offline)} -> {len(DNase_offline_DNADiffusion)}")
print(DNase_complete_DNADiffusion["TAG"].value_counts())
print(DNase_offline_DNADiffusion["TAG"].value_counts())

DNase_complete_DNADiffusion.to_csv("/home/wangyucheng/projects/SeqEval_Bench_Data/DNase/CellType/DNase_complete_starting_sequences_DNADiffusion.tsv", sep="\t", index=False)
DNase_offline_DNADiffusion.to_csv("/home/wangyucheng/projects/SeqEval_Bench_Data/DNase/CellType/DNase_offline_starting_sequences_DNADiffusion.tsv", sep="\t", index=False)
```
