# DNase

```python
DNase_complete = pd.read_csv("/home/wangyucheng/projects/SeqEval_Bench_Data/DNase/CellType/DNase_complete_starting_sequences.tsv", sep="\t")
DNase_offline = pd.read_csv("/home/wangyucheng/projects/SeqEval_Bench_Data/DNase/CellType/DNase_offline_starting_sequences.tsv", sep="\t")
```
