```python
from pathlib import Path

# =========================
# main
# =========================
records = []
for file_path in complete_path:
    p = Path(file_path)
    dataset, cell_type = p.parent.parent.name, p.parent.name
    tag = f"{dataset}_{cell_type}"

    scores = pd.read_csv(file_path, sep="\t", usecols=["functional_score"])["functional_score"].dropna()
    records.append({
        "dataset": dataset,
        "cell_type": cell_type,
        "dataset_tag": tag,
        "n_rows": len(scores),
        "min": scores.min(),
        "max": scores.max(),
    })

minmax_df = pd.DataFrame(records)

minmax_df
```
